```python
import pandas as pd
```

```python
elig = pd.read_csv("../Data/Raw/eligibility_processing.csv")
```

```python
md_elig = elig[elig["State Name"] == "Maryland"].copy()
```

```python
md_elig["priority"] = md_elig["Original or Updated"].map({"O": 0, "U": 1})

md_elig = (
    md_elig
    .sort_values(["Reporting Period", "priority"])
    .drop_duplicates("Reporting Period", keep="last")
    .drop(columns="priority")
)
```

```python
md_elig = md_elig[
    [
        "Reporting Period",
        "Original or Updated",
        "Beneficiaries with a Renewal Due",
        "Beneficiaries Whose Coverage Was Renewed (Total)",
        "Beneficiaries Disenrolled at Renewal (Total)",
        "Beneficiaries Determined Ineligible at Renewal",
        "Beneficiaries Disenrolled for Procedural Reasons at Renewal",
        "Beneficiaries with a Pending Renewal"
    ]
].dropna().copy()
```

```python
md_elig["Month"] = pd.to_datetime(
    md_elig["Reporting Period"].astype(str),
    format="%Y%m"
)

md_elig.to_csv(
    "../Data/Processed/eligibility_processing_clean.csv",
    index=False
)
```
