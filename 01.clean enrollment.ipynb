```python
import pandas as pd
```

```python
df = pd.read_csv("../Data/Raw/enrollment.csv")
```

```python
md = df[df["State Name"] == "Maryland"].copy()
```

```python
md = md[
    [
        "Reporting Period",
        "Preliminary or Updated",
        "Final Report",
        "Total Medicaid Enrollment",
        "Total CHIP Enrollment",
        "Total Medicaid and CHIP Enrollment"
    ]
].copy()
```

```python
md["priority"] = md["Preliminary or Updated"].map({"P": 0, "U": 1})

md = (
    md
    .sort_values(["Reporting Period", "priority"])
    .drop_duplicates("Reporting Period", keep="last")
    .drop(columns="priority")
)
```

```python
md = md.dropna(
    subset=[
        "Total Medicaid Enrollment",
        "Total CHIP Enrollment"
    ]
).copy()
```

```python
md["Month"] = pd.to_datetime(
    md["Reporting Period"].astype(str),
    format="%Y%m"
)
```

```python
md.to_csv(
    "../Data/Processed/enrollment_clean.csv",
    index=False
)
```
